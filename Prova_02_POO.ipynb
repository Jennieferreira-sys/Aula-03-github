3 - Implemente um programa em Python utilizando orientação a objetos para simular a impressão de
documentos.

Crie uma classe Documento com atributos privados título, conteúdo e tipo, onde o tipo pode representar
categorias como "relatório", "carta", "artigo", entre outros.

Implemente métodos públicos para acessar essas informações de forma segura.
Em seguida, implemente uma classe Impressora com o método imprimir, que recebe um objeto da classe
Documento e exibe suas informações na tela. A impressora deve apenas utilizar o documento no momento
da impressão, sem manter uma referência permanente a ele, caracterizando uma relação de dependência.

O programa principal deve apresentar um menu interativo no console que permita ao usuário criar
documentos com diferentes tipos, visualizar seus conteúdos por meio da impressora e encerrar o programa.

Utilize encapsulamento para proteger os dados dos documentos e boas práticas de organização orientada a
objetos.

```python
class Documento: #Classe documentos#
    def __init__(self, titulo, conteudo, tipo): # Atributos privados#
        self.__titulo = titulo
        self.__conteudo = conteudo
        self.__tipo = tipo

    def get_titulo(self): # tornando esses métodos públicos para acessar as informações.(Devolvendo seu valor)#
        return self.__titulo

    def get_conteudo(self):
        return self.__conteudo

    def get_tipo(self):
        return self.__tipo


class Impressora:
    def imprimir(self, documento): #impressora recebe documentos#
        print("\n===== DOCUMENTO =====")
        print("Título:", documento.get_titulo())
        print("Tipo:", documento.get_tipo())
        print("Conteúdo:", documento.get_conteudo())
        print("=====================")


# Programa principal

documento = None
impressora = Impressora()

#Menu#
while True:
    print("\n===== SISTEMA DE IMPRESSÃO =====")
    print("1 - Criar documento")
    print("2 - Imprimir documento")
    print("3 - Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        titulo = input("Digite o título: ")
        conteudo = input("Digite o conteúdo: ")

        print("\nTipos de documento:")
        print("1 - Relatório")
        print("2 - Carta")
        print("3 - Artigo")
        print("4 - Outro")

        tipo_opcao = input("Escolha o tipo: ")

        if tipo_opcao == "1":
            tipo = "Relatório"
        elif tipo_opcao == "2":
            tipo = "Carta"
        elif tipo_opcao == "3":
            tipo = "Artigo"
        elif tipo_opcao == "4":
            tipo = "Outro"
        else:
            print("Opção de tipo inválida.")
            tipo = "Outro"

        documento = Documento(titulo, conteudo, tipo)

        print("\n Documento criado com sucesso!")

    elif opcao == "2":
        if documento is None:
            print("\n Nenhum documento foi criado.")
        else:
            impressora.imprimir(documento)

    elif opcao == "3":
        print("\n Programa encerrado.")
        break

    else:
        print("\n Opção inválida. Tente novamente.")
```
